### Why Acuuracy Is Misleading for Imbalanced Fraud Detection -
Accuracy can be misleading metric when evalating fraud detection models because the dataset is highly imbalanced. In this dataset, approximately 99.83% of transaction are legitimate, while only 0.17% are fradulent.

For example, a model that predicts every transaction as legitimate would achieve approximately 99.83% accuracy.However, it would detect **0% of the fraudulent transaction**,making the model useless for fraud detection.

Therefore, accuracy alone is not sufficient for evaluating an imbalanced fraud detection model. Metrics such as **precision, recall, F1-score,ROC-AUC, and PR-AUC** should also be considered.

**Precision:** Measures how many transactions predicted as fraud are actually fraudulent.
**Pecall:** Measures how many actual transactions are correctly detected.
**F1-score:** provides a balance between precision and recall.
**ROC-AUC:** Measures the model's ability to distinguish between fraud and non-fraud transations.
**PR-AUC:** Is particularly useful for hihghly imbalanced datasets because it focuses on the performance of the minority class.

For fraud detection, recall is specially important because failing to identify a fraudulent transaction can result in financial loss.

# Handling Class Immbalance Using SMOTE
The dataset contains a severe class imbalance, with fraudulent transactions representing only a small percentage of all transactions. To address this problem, SMOTE was applied to the training data.

SMOTE creates synthetic samples of the minority class (fraudulent transactions) based onexisting minority-class observations.This increases the representation of fraudulent transactions and helps the machine learning model learn pattern associated with fraud.

SMOTE was appilied only to the training data after spltting the dataset into training sets. The test set was kept unchanged to prove an unbiased evaluation to the model's performance.

# Most Important Metric for Fraud Detection: Recall vs Precision
For credit fraud detection , **Recall is generally more important than precision** because missing a fraudulent transaction can result in financial loss for the customer and the bank.

**Recall** measures the perentage of actual fraudulent transactions that the model correctly identifies. A high recall means that the model is able to detect most of the fraudulent transactions.

**Precision** measures the percentage of transactions predicted as fraud that are actually fraudulent. A low precision means that the model may incorrectly flag many legitimate transactions as fraud.

There is a trade-off between Recall and Precision:
-**High Recall:** Detect more fraudulent transactions but may also flag more legitimate transactions as fraud.
-**High Precision:** Produces fewer false fraud alerts but may miss some actual fraudulent transactions.

For a fraud detection system, a **high Recall is usually preferred**, because failing to detect fraud (False Negative) can be more costly than temporarily flagging a legitimate transaction for verification (False Positive).

However, Precision should not be ignored. A model with extremely low precision could generate too many false alarms, causing inconvenience to customers and increasing the workload for fraud investigation teams.

Therefore, the ideal fraud detection model should achieve a **good balance between Recall and Precision**, which can be measured using the **F1-Score**.

# Scalability: Handling 1 million Transactions per Hour
The current fraud detection model was trained on a relatively smaller dataset, but real-world payment system may need to process around 1 million transactions per hour. This is approximately 278 transactions per second.

To handle this scale, the model would need an efficient and scalable deployment architecture.

-**Efficient preprocessing:** Feature preprocessing should be optimized and performed quickly so that it does not become a bottleneck.
-**Model optimization:** A lightweight model such as Logistic Regression can provide very fast predictions. Random Forest can also be optimized by controlling the number and depth of trees.
-**Batch processing:** Transactions can be processed in small batches instead of one at a time when real-time processing is not required.
-**Parallel processing:** Multiple CPU cores or distributed computing can process transactions simultaneously.
-**Horizontal scaling:** Multiple instances of fraud detection service can run in the parallel behind a load balancer to distribute incoming transactions.
-**Real-time monitoring:** The system should monitor prediction latency, throughout, memory usage, and model performance.
-**Model retraining:** The model should be periodically retained using new transaction and fraud data so that it can adapt to changing fraud patterns.

At 1 million transactions per hour, the system would need to process approximately 278 transactions per second. A production system should therefore be benchmarked under realistic load before deployment.

Overall, the trained model itself may be suitable for high-throughout prediction, but handling 1 million transactions per hour would require optimized preprocessing, parallel or distributed infrastructure, monitoring and a production-ready deployment architecture.
