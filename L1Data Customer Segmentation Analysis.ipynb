# Descriptive statistics

```python
avarage_purchase_value=data["Total Purchase Amount"].mean()
print("Average Purchse Value:", round(avarage_purchase_value,2))
```

```python
purchase_frequency=(
    data.groupby("Customer ID")
    .size()
    .reset_index(name='Purchase Frequency')
)
purchase_frequency.head()
```

```python
average_purchase_frequency=purchase_frequency['Purchase Frequency'].mean()
print("Average Purchase Frequency:",round(average_purchase_frequency,2))
```

```python
data['Purchase Date'] = pd.to_datetime(data['Purchase Date'])

customer = data.groupby('Customer ID').agg(
    total_purchase=('Total Purchase Amount', 'sum'),
    purchase_frequency=('Customer ID', 'count'),
    first_purchase=('Purchase Date', 'min'),
    last_purchase=('Purchase Date', 'max')
).reset_index()

customer['average_purchase_value'] = (
    customer['total_purchase'] /
    customer['purchase_frequency']
)

customer['lifespan_years'] = (
    (customer['last_purchase'] - customer['first_purchase']).dt.days
    / 365.25
)


average_lifespan = customer.loc[
    customer['lifespan_years'] > 0,
    'lifespan_years'
].mean()


customer['annual_purchase_frequency'] = np.where(
    customer['lifespan_years'] > 0,
    customer['purchase_frequency'] /
    customer['lifespan_years'],
    customer['purchase_frequency']
)


customer['CLV'] = (
    customer['average_purchase_value'] *
    customer['annual_purchase_frequency'] *
    average_lifespan
)


print("Average Purchase Value:",
      round(customer['average_purchase_value'].mean(), 2))

print("Average Purchase Frequency:",
      round(customer['purchase_frequency'].mean(), 2))

print("Average Customer Lifespan:",
      round(average_lifespan, 2), "years")

print("Average Customer Lifetime Value:",
      round(customer['CLV'].mean(), 2))
```

#  Feature selection

```python
analysis_date = data['Purchase Date'].max() + pd.Timedelta(days=1)


rfm = data.groupby('Customer ID').agg(
    Recency=('Purchase Date',
             lambda x: (analysis_date - x.max()).days),

    Frequency=('Customer ID', 'count'),

    Monetary=('Total Purchase Amount', 'sum')
).reset_index()


rfm.head()
```

# Data normalisation/standardisation

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
numeric_cols=data.select_dtypes(include=['int64','float64']).columns
```

```python
scaler=StandardScaler()
x_scaled=scaler.fit_transform(data[numeric_cols])
x_scaled_df=pd.DataFrame(
    x_scaled,
    columns=numeric_cols)
print("Shape:",x_scaled_df.shape)
print("\nMean:")
print(x_scaled_df.mean().round(4))
print('\nStandard deviation:')
print(x_scaled_df.std(ddof=0).round(4))
```

#  K-Means clustering 

```python
import matplotlib.pyplot as plt
```

```python
wcss=[]
for k in range(2,11):
    kmeans=KMeans(
        n_clusters=k,random_state=42,n_init=10)
    kmeans.fit(x_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8,5))
plt.plot(range(2,11),wcss,marker='o')
plt.xlabel('Number of Cluster (k)')
plt.ylabel('WCSS / Inertia')
plt.title('Elbow Method for Optimal k')
plt.xticks(range(2,11))
plt.show()
```

```python
optimal_k=4
kmeans=KMeans(n_clusters=optimal_k,random_state=42,n_init=10)
data['Cluster']=kmeans.fit_predict(x_scaled)
data.head()

data['Cluster'].value_counts().sort_index()

```

```python
features=numeric_cols
plt.figure(figsize=(8,6))
plt.scatter(
    x_scaled[:,0],
    x_scaled[:,1],
    c=data['Cluster'],cmap='viridis',alpha=0.6)
plt.xlabel(features[0])
plt.xlabel(features[1])
plt.title(f'K-Means Clusters: {features[0]} vs {features[1]}')
plt.colorbar(label='Cluster')
plt.show()
```

```python
features=numeric_cols
plt.figure(figsize=(8,6))
plt.scatter(
    x_scaled[:,2],
    x_scaled[:,3],
    c=data['Cluster'],cmap='viridis',alpha=0.6)
plt.xlabel(features[2])
plt.xlabel(features[3])
plt.title(f'K-Means Clusters: {features[2]} vs {features[3]}')
plt.colorbar(label='Cluster')
plt.show()
```
