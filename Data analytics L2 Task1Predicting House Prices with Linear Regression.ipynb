# Feature Selection Discussion

The target variable is price .

**Area**:Larger houses generally have higher prices becauses they provide more living space.
**Bedrooms**:The number of bedrooms can affect the value and demand of a house.
**Bathrooms**:Houses with more bathrooms are generally more valuable .
**Floors**:The number of floors can influence the size and value of a property.
**YearBuilt**:Newer houses may have higher prices due to modern construction and facilities.
**Location**:Location is an important factor because property prices can vary between different areas.
**Condition**:Houses in better condition are generally more valuable.
**Garage**:A garage can increase property value by providing parking and additional utility.


```python
categorical_features =["Location",'Condition','Garage']
df_encoded =pd.get_dummies(
    df,
    columns=categorical_features,
    drop_first=True
)
df_encoded.head()
```

```python
print(df_encoded.head())
```

```python
df_encoded = df_encoded.drop('Id',axis=1)
```

```python
x=df_encoded.drop("Price",axis=1)
y=df_encoded['Price']
x.shape
```

```python
y.shape
```

```python
correlation_matrix=df_encoded.corr()
plt.figure(figsize=(10,6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

```python
price_correlation=correlation_matrix['Price'].sort_values(ascending=False)
print(price_correlation)
```

# Correlation Analysis
The correlation heatmap shows the relationship between the  numerical / encoded features and the marget variable,**Price**
Feature with correlation values closer to **+1** have a strong positive relationship with house price, while values closer to **-1** have strong negative relationship. values close to **0** indicate a weak linear relationship.
The features with the highest absolute correlation with **Price** are considered the most important potential predictors.However, correlation does not imply causation.

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.20,random_state=42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
model =LinearRegression()
model.fit(x_train,y_train)
```

```python
print(model.intercept_)
coefficients =pd.DataFrame({'Feature':x_train.columns,'Coefficient':model.coef_})
print(coefficients)
```

```python
y_pred = model.predict(x_test)
print(y_pred[:10])
```

```python
from sklearn.metrics import mean_squared_error,r2_score
```

```python
mse=mean_squared_error(y_test,y_pred)
rmse=np.sqrt(mse)
r2=r2_score(y_test,y_pred)
print(mse)
print(rmse)
print(r2)
```

```python
result=pd.DataFrame({'Metric':['MSE','RMSE','R2 Score'],'Vaiue':[mse,rmse,r2]})
result
```

```python
plt.figure(figsize=(8,6))
plt.scatter(y_test,y_pred,alpha=0.6)
plt.plot(
    [y_test.min(),y_test.max()],[y_test.min(),y_test.max()],linestyle="--")
plt.xlabel("Actual Prices")
plt.ylabel('Predicted Prices')
plt.title('Actual vs Predicted House Prices')
plt.show()
```

```python
residuals=y_test - y_pred
plt.figure(figsize=(8,6))
plt.scatter(y_pred,residuals,alpha=0.6)
plt.axhline(y=0,linestyle="--")
plt.xlabel('Predicted Prices')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.show()
```

```python
coefficients =pd.DataFrame({'Feature':x_train.columns,'Coefficient':model.coef_})
coefficients_sorted = coefficients.sort_values(by='Coefficient',ascending=False)
print(coefficients_sorted)
```

```python
print(coefficients_sorted.head(5))
print(coefficients_sorted.tail(5))
```

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error , r2_score
```

```python
ridge_model= Ridge(alpha=1.0)
ridge_model.fit(x_train,y_train)
```

```python
ridge_pred=ridge_model.predict(x_test)
```

```python
ridge_mse=mean_squared_error(y_test,ridge_pred)
ridge_rmse=np.sqrt(ridge_mse)
ridge_r2=r2_score(y_test,ridge_pred)
print(ridge_mse)
print(ridge_rmse)
print(ridge_r2)
```

```python
comparison =pd.DataFrame({
    'Model':['Linear Regression','Ridge Regression'],
    'MSE':[mse,ridge_mse],
    'RMSE':[rmse,ridge_rmse],
    'R2 Score':[r2,ridge_r2]})
comparison
```
