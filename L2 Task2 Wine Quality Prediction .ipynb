# Class Imbalance
Class imbalance is present in the Wine Quality dataset. Quality scores 5 and 6 are much frequent, while extreme scores such as 3 and 8 are underrepresented. This imbalance may cause a classification model to favor the majority classes and perform poorly on minority classes. Therefore, accuracy alone may not be sufficient for evaluation; precision, recall, F1-score, and a xonfution matrix should  also be considered.

#  quality scores

```python
def quality_group(score):
    if score <= 4:
        return 'Low'
    elif score <=6:
        return 'Medium'
    else:
        return 'High'

df['quality_group']=df['quality'].apply(quality_group)

print(df[['quality','quality_group']].head(10))

print('\nClass distribution:')
print(df['quality_group'].value_counts())
```

```python
print(df['quality_group'].value_counts(normalize=True).mul(100).round(2))
```

```python
df['quality_group'].value_counts().reindex(['Low','Medium','High']).plot(kind='bar')
plt.xlabel=('Quality Group')
plt.ylabel('Number of Wines')
plt.title('Wine Quality Group')
plt.xticks(rotation=0)
plt.show()
```

```python
df['quality_binary']=df['quality'].apply(
    lambda x: 'Good' if x >=6 else 'Bad')
print(df['quality_binary'].value_counts())
```

# Justification
i chose a 3-class classification approach because it provides a balance between retaining information and reducing class imbalance. The quality scores were grouped into low(<=4),medium(5-6), and high(>=7). This is more informative than simply classifying wines as good or bad, while making the prediction problem easier than predicting every individual quality score.

# Train/test split with stratification

```python
from sklearn.model_selection import train_test_split
x=df.drop(columns=['quality','quality_group'])
y=df['quality_group']
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.20,random_state=42,stratify=y)
print('Training set:',x_train.shape)
print('Testing set:',x_test.shape)
```

```python
print('Original class distribution:')
print(y.value_counts(normalize=True).round(3))

print('\nTraining class distribution:')
print(y_train.value_counts(normalize=True).round(3))

print('\nTesting class distribution:')
print(y_test.value_counts(normalize=True).round(3))
```

# Justification
The dataset was didided into 80% taining and 20% testing sets using stratified sampling. Stratification was used to preserve the propertion low, medium, and high quality classes in both sets. This is important because the dataset has class imbalance, and an ordinary random split could produce different class propertions in the training and testing sets.

# Random Forest, SGD, and SVC

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import SGDClassifier
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model= RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight='balanced')
rf_model.fit(x_train,y_train)
```

```python
x=df.drop(columns=['quality','quality_group','quality_binary'])
y=df['quality_group']

print('x columns:')
print(x.columns.tolist())

print('\nx data types:')
print(x.dtypes)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.20,random_state=42,stratify=y)
```

```python
scaler = StandardScaler()
print('Has transform',hasattr(scaler,'transform'))
print('method',[x for x in dir(scaler) if "transform" in x])
```

```python
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
sgd_model = SGDClassifier(
    max_iter=1000,random_state=42,class_weight='balanced')
sgd_model.fit(x_train_scaled,y_train)
```

```python
svc_model= SVC(
    kernel='rbf',class_weight='balanced',random_state=42)
svc_model.fit(x_train_scaled,y_train)
```

# accuracy, classification report, confusion matrix

```python
from sklearn.metrics import accuracy_score, classification_report
y_pred_rf = rf_model.predict(x_test)

print('========= RANDOM FOREST ==========')
print('Accuracy:', accuracy_score(y_test,y_pred_rf))
print('\nClassification Report:')
print(classification_report(y_test,y_pred_rf))
```

```python
y_pred_sgd=sgd_model.predict(x_test_scaled)
print('========= SGD ==========')
print('Accuracy:', accuracy_score(y_test,y_pred_sgd))
print('\nClassification Report:')
print(classification_report(y_test,y_pred_sgd))
```

```python
y_pred_svc=svc_model.predict(x_test_scaled)
print('========= SVC ==========')
print('Accuracy:', accuracy_score(y_test,y_pred_svc))
print('\nClassification Report:')
print(classification_report(y_test,y_pred_svc))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
```

```python
cm_rf = confusion_matrix(y_test,y_pred_rf,labels=['Low','Medium','High'])
disp = ConfusionMatrixDisplay(confusion_matrix=cm_rf , display_labels=['Low','Medium','High'])
disp.plot()
plt.title('Random Forest - Confusion Matrix')
plt.show()
```

```python
cm_sgd = confusion_matrix(y_test,y_pred_sgd,labels=['Low','Medium','High'])
disp = ConfusionMatrixDisplay(confusion_matrix=cm_sgd , display_labels=['Low','Medium','High'])
disp.plot()
plt.title('SGD - Confusion Matrix')
plt.show()
```

```python
cm_svc = confusion_matrix(y_test,y_pred_svc,labels=['Low','Medium','High'])
disp = ConfusionMatrixDisplay(confusion_matrix=cm_svc , display_labels=['Low','Medium','High'])
disp.plot()
plt.title('SVC - Confusion Matrix')
plt.show()
```

#  performance of all 3 models

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score , f1_score
```

```python
comparison = pd.DataFrame({'Mode':['Random Forest','SGD','SVC'],
                           'Accuracy':[accuracy_score(y_test,y_pred_rf),accuracy_score(y_test,y_pred_sgd),accuracy_score(y_test,y_pred_svc)],
                           'Precision':[precision_score(y_test,y_pred_rf,average='weighted'),precision_score(y_test,y_pred_sgd,average='weighted'),precision_score(y_test,y_pred_svc,average='weighted')],
                           'Recall':[recall_score(y_test,y_pred_rf,average='weighted'),recall_score(y_test,y_pred_sgd,average='weighted'),recall_score(y_test,y_pred_svc,average='weighted')],
                           'F1 Score':[f1_score(y_test,y_pred_rf,average='weighted'),f1_score(y_test,y_pred_sgd,average='weighted'),f1_score(y_test,y_pred_svc,average='weighted')],
                          })
comparison.iloc[:,1:]=comparison.iloc[:,1:].round(3)
display(comparison)
```

```python
best_model = comparison.loc[comparison['F1 Score'].idxmax(),'Mode']
print('Best model based on weighted F1 Score:', best_model)
```

# Conclusion :
Among the three classifier, the model the highest weighted F1-score is the most suitable for deployment.
The weigghted F1-score is considerd important because the wine quality classes are imbalanced.
The selected model provides the best overall balance between precision and recall while handling the different quality classes.
Therefore, Random Forest would be the preferred model for deployment.
